```julia
using Plots
gr()
using DelimitedFiles
#==========getting the data============#
Data = DelimitedFiles.readdlm("EVDData.csv",',')
Epidays = Data[:,1]
TotCases = Data[:,2]
#==========function for the SIR========#
function updateSIR(popnvector)
    S = popnvector[1]
    I = popnvector[2]
    R = popnvector[3]
    NewS = S - lambda*I*S*dt
    NewI = I + lambda*I*S*dt - gamma*I*dt
    NewR = R + gamma*I*dt
    return [NewS NewI NewR]
    
end

```

```julia
lambda = 1.47*10^-6.
gamma = 1/8
dt = 0.5
tfinal= 610.

S0 = 10^5
I0 = 20.
R0 = 0.

nsteps = round(Int64, tfinal/dt)
resultvals = Array{Float64}(undef,(nsteps+1),3)
timevec = Array{Float64}(undef,nsteps+1)
resultvals[1,:] = [S0 I0 R0]
timevec[1] = 0.  

for step = 1:nsteps
    resultvals[step+1,:] = updateSIR(resultvals[step,:])
    timevec[step+1] = timevec[step]+dt
end
Ivals = resultvals[:,2]
Rvals = resultvals[:,3]
Cvals = Ivals + Rvals

plot(timevec, Cvals,
label = "Model values",
xlabel = "Epidemic Days",
ylabel = "Number of cases to date",
title = "Model versus data"
)
plot!(Epidays,TotCases,
legend = :right,
line = :scatter,
label = "Reported number of cases"
)
```
